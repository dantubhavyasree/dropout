```
import torch
from src.dataset import mnist_loaders, device
from src.model import Net
from src.utils import predict, show_samples

device_ = device()
_, test_loader, _ = mnist_loaders(batch_size=128)

model = Net(dropout=True).to(device_)
model.load_state_dict(torch.load('models/mnist_dropout.pt', map_location=device_))
model.eval()
```

```
# take the first 3 test images as samples
images, labels = next(iter(test_loader))
images, labels = images[:3], labels[:3]

probs = predict(model, images, device_)
preds = [(p.argmax().item(), p.max().item()) for p in probs]
for k in range(3):
    show_samples(images[k:k+1], [labels[k].item()], [preds[k]],
                f'results/sample_results/result_0{k + 1}.png')
    print(f'sample {k + 1}: true={labels[k].item()}, pred={preds[k]}')
```

The images above are saved under `results/sample_results/` so they can
be committed to the repository together with the plots and the model.
